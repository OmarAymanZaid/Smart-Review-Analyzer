# Notebook 1 — Preprocessing
Cleans raw Amazon Reviews and writes `data/processed/clean_reviews.csv`.
Run this **once** on Kaggle before training any model.
The output CSV is the single source of truth for all three models.

```python
# ── When running on Kaggle, add project root to sys.path ─────────────────
import sys
from pathlib import Path

# Adjust if your notebook is nested differently
PROJECT_ROOT = Path('/kaggle/working/sentiment_project')
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

# Override CLEAN_CSV path for Kaggle working dir
KAGGLE_RAW_CSV  = Path('/kaggle/input/amazon-reviews-dataset/Amazon_Reviews.csv')
KAGGLE_OUT_CSV  = PROJECT_ROOT / 'data' / 'processed' / 'clean_reviews.csv'
```

```python
import pandas as pd
from src.preprocessing import preprocess_dataset

print('Loading raw data...')
df_raw = pd.read_csv(KAGGLE_RAW_CSV, engine='python')
print(f'Raw shape: {df_raw.shape}')

print('Preprocessing...')
df_clean = preprocess_dataset(df_raw)
print(f'Clean shape: {df_clean.shape}')
df_clean.head()
```

```python
# Sentiment distribution
print(df_clean['Sentiment'].value_counts())
print(df_clean['Sentiment'].value_counts(normalize=True).mul(100).round(1))
```

```python
KAGGLE_OUT_CSV.parent.mkdir(parents=True, exist_ok=True)

# Keep only the columns needed downstream
df_out = df_clean[['Cleaned Review', 'Sentiment']].rename(
    columns={'Cleaned Review': 'text', 'Sentiment': 'label'}
)
df_out.to_csv(KAGGLE_OUT_CSV, index=False)
print(f'Saved {len(df_out):,} rows → {KAGGLE_OUT_CSV}')
```
